# 04. Anomaly Detection & Marketing Recommendations
## Questions 3 & 5: Which posts are unusual & what should marketing do next?

Isolation Forest normalizes engagement ratios within each content type so posts are compared strictly against comparable peers.

```python
import sys; sys.path.append('../src')
import pandas as pd
import engine as E

posts, comments = E.load_data('../data')
anom = E.find_anomalies(posts, contamination=0.04)
outliers = anom[anom['anomaly']].sort_values(by='anomaly_score', ascending=False)
print(f'Total Anomalies Flagged: {len(outliers)}')
display(outliers[['post_id', 'brand', 'content_type', 'views', 'likes', 'comments_count', 'engagement_rate', 'anomaly_score']].head(8))
```

```python
# Generate Recommendations
perf = E.content_performance(posts)
clustered_c = E.cluster_comments(comments, k=6)
recs = E.generate_recommendations(clustered_c, perf)

print('Evidence-Based Marketing Actions:')
for r in recs:
    print('-', r)
```
