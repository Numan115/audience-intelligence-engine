Character $n$-grams (`char_wb`, range 2–4) combined with Roman Urdu stopword cleaning allow the model to cluster variations like *kitna*, *kitnaa*, and *ketna* without manual labeling.

```python
import sys; sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import engine as E

posts, comments = E.load_data('../data')
clustered_c = E.cluster_comments(comments, k=6)

print('Discovered Cluster Counts:')
print(clustered_c['cluster_name'].value_counts())
```

```python
# Crosstabulation: Clusters vs. Language Type
ct = pd.crosstab(clustered_c['cluster_name'], clustered_c['language_type'])
display(ct)
```
