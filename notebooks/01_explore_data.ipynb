# 01. Exploratory Data Inspection
## Pakistan Social Audience & Campaign Intelligence Engine

### Overview
In this notebook, we load and validate the two relational tables:
- **Table A: Social Media Posts (`posts.csv`)** - 2,000 public post records across Pakistani fashion brands.
- **Table B: Comments (`comments.csv`)** - 8,000 customer comments containing Roman Urdu and English interactions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
posts = pd.read_csv('../data/posts.csv')
comments = pd.read_csv('../data/comments.csv')

print(f'Posts shape: {posts.shape}')
print(f'Comments shape: {comments.shape}')
display(posts.head(3))
display(comments.head(3))
```

```python
# Brand and Platform Breakdown
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
posts['brand'].value_counts().plot(kind='bar', ax=axes[0], color='#2b5c8f')
axes[0].set_title('Posts by Pakistani Fashion Brand', fontweight='bold')
posts['platform'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%', colors=['#4ca585', '#d4a373', '#e76f51'])
axes[1].set_title('Posts by Platform', fontweight='bold')
axes[1].set_ylabel('')
plt.tight_layout()
plt.show()
```

```python
# Comment Language Breakdown
print('Language distribution in comments:')
print(comments['language_type'].value_counts(normalize=True) * 100)
```
